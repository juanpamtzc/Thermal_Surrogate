```python
# Cell 2
import glob

# Find all run directories in the openfoam/runs folder
run_dirs = sorted(glob.glob("openfoam/runs/run_*"))

print(f"Found {len(run_dirs)} completed simulation runs:\n")

for run in run_dirs:
    # Check if the OpenFOAM log file exists to confirm it actually ran
    log_file = os.path.join(run, "solver.log")
    if os.path.exists(log_file):
        print(f"✅ {os.path.basename(run)} - Successfully completed (Log found)")
    else:
        print(f"❌ {os.path.basename(run)} - Failed or missing log")
```
